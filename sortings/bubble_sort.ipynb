```python
def bubble(arr):
    n = len(arr)

    for i in range(n-1, -1, -1):
        for j in range(0, i):
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]
        
    print(*arr)

arr1 = [54,23,5,2,1]
bubble(arr1)

# T.C - O(N2)
# S.C - O(1)
```

```python
def bubbles(arr):
    n = len(arr)
    for i in range(n-1, -1, -1):
        for j in range(0, i):
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]

    print(*arr)

arr = [5,4,2,234,1]
bubbles(arr)
```

```python
def bubble_optimal(arr):
    n = len(arr)
    for i in range(n-1, -1, -1):
        did_swap = False
        for j in range(0, i):
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]
                did_swap = True
        if not did_swap:
                break
        

    print(*arr)

arr = [1,2,3,4,5]
bubble_optimal(arr)

# T.C (WORST CASE and AVERAGE CASE) - O(N2)
# T.C (BEST CASE) - O(N)
# S.C - O(1)
```
