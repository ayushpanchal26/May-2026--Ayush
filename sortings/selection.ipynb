```python
def selection(arr):
    n = len(arr)
    for i in range(n-1):
        min_index = i 
        for j in range(i+1, n):
            if arr[j]< arr[min_index]:
                min_index = j
                
            arr[i], arr[min_index] = arr[min_index], arr[i]

    
    print(*arr)


arr = [12,46,24,52,20,9]
selection(arr)

        
```
