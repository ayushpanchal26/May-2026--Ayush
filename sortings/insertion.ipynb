```python
def insertion(arr):
    n = len(arr)
    for i in range(n):
        j = i
        while(j>0 and arr[j-1] > arr[j]):
            arr[j-1], arr[j] = arr[j], arr[j-1]
            j-=1

    print(*arr)
    
arr = [5,4,3,2,1]
insertion(arr)
```
