```python
def bubble_sort(arr, n):
    if n == 1:
        return
    
    for j in range(n-1):
        if arr[j] > arr[j+1]:
            arr[j], arr[j+1] = arr[j+1], arr[j]

    bubble_sort(arr, n-1)


    
arr = [5,4,3,2,1]
n = len(arr)

print("Before using Bubble sort with recursion")
print(*arr)

bubble_sort(arr, n)
print("After using bubble sort")
print(*arr)

# T.C - O(N^2)
# S.C - O(N)
```

```python
def bubble_sort(arr, n):
    if n == 1:
        return
    did_swap = False
    for j in range(n-1):
        if arr[j] >arr[j+1]:
            arr[j], arr[j+1] = arr[j], arr[j+1]
            did_swap = True

    if not did_swap:
        print("only runs when array is Already sorted")
        return

    bubble_sort(arr, n-1)


arr = [1,2,3,4,5]
bubble_sort(arr, len(arr))
print

# T.C - O(N^2) for worst and average case but for Best case - O(N)
# S.C - O(N)
```
