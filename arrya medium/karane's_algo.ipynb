Brute

```python
def kardanes(arr):
    maxi = float('-inf')
    for i in range(len(arr)):
        for j in range(i, len(arr)):
            sum = 0
            for k in range(i, j+1):
                sum += arr[k]
            maxi = max(maxi, sum)

    return maxi

arr = [2,3,5,-2,7,-4]
kardanes(arr)

# T.C - O(N^3)
# S.C - O(1)
```

Better

```python
def better(arr):
    maxi = float('-inf')

    for i in range(len(arr)):
        sum = 0 
        for j in range(i, len(arr)):
            sum += arr[j]
            maxi = max(maxi, sum)

    return maxi

arr = [2,3,5,-2,7,-4]
better(arr)

# T.C - O(N^2)
# S.C - O(1)
```
