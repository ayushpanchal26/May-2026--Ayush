Brute

```python
def brute(arr):
    n = len(arr)
    pos = []
    neg = []
    for i in range(n):
        if arr[i] <=0: neg.append(arr[i])
        else: pos.append(arr[i])

    for i in range(n//2):
        arr[2*i] = pos[i]
        arr[2*i+1] = neg[i]

    return arr

arr = [1,2,-4,-5]
brute(arr)

# T.C - O(N + N/2)
# S.C - O(N)
```

```python
def optimal(arr):
    n = len(arr)

    ans = [0] * n

    pos_index = 0
    neg_index = 1

    for i in range(n):
        if arr[i] > 0:
            ans[pos_index] = arr[i]
            pos_index +=2

        else:
            ans[neg_index] = arr[i]
            neg_index +=2

    return ans


arr = [1, 2, -4, -5]
optimal(arr)

# T.C - O(N)
# S.C - O(N)
```
