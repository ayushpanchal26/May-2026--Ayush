Brute

```python
def my_brute(arr):
    max_profit  =0 

    for i in range(len(arr)):
        for j in range(i+1, len(arr)):
            profit = arr[j] - arr[i]
            max_profit = max(max_profit, profit)

    return max_profit

arr = [7,1,5,3,6,4]
my_brute(arr)

# T.C - O(N^2)
# S.C - O(1)
```

Optimal

```python
def optimal(arr):
    min_prices = float('inf')
    max_profit = 0

    for i in arr:
        if i < min_prices:
            min_prices = i
        else:
            max_profit = max(max_profit, i - min_prices)

    return max_profit


arr = [7,1,5,3,6,4]
optimal(arr)
```
