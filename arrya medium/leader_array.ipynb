Brute

```python
def brute(arr):
    n = len(arr)
    ans = []

    for i in range(n):
        leader = True
        for j in range(i+1, n):
            if arr[j] >= arr[i]:
                leader = False
                break

        if leader:
            ans.append(arr[i])
    
    return ans

arr = [4, 7, 1, 0]
brute(arr)

# T.C - O(N^2)
# S.C - O(1)
```

Optimal

```python
def optimal(arr):
    ans = []

    if not arr:
        return ans
    
    max_val = arr[-1]
    ans.append(max_val)

    for i in range(len(arr) - 2, -1, -1):
        if arr[i] > max_val:
            ans.append(arr[i])
            max_val = arr[i]

    ans.reverse()

    return ans

arr = [10, 22, 12, 3, 0, 6]
optimal(arr)
```
