Brute

```python
def my_brute(arr):
    a = 0
    b = 0 
    c = 0

    for i in range(len(arr)):
        if arr[i] == 0:
            a+=1
        elif arr[i] == 1:
            b+=1
        else:
            c+=1
    
    index = 0

    for i in range(a):
        arr[index] = 0
        index +=1

    for i in range(b):
        arr[index] = 1
        index +=1

    for i in range(c):
        arr[index] = 2
        index+=1

nums = [1,0,2,1,0]
my_brute(nums)
print(nums)

# T.C - O(N)
# S.C - O(1)
```

```python
def better(arr):
    a = b = c = 0

    for i in arr:
        if i == 0:
            a +=1
        elif i == 1:
            b+=1
        else:
            c+=1
        
    for i in range(a):
        arr[i] = 0
    
    for i in range(a, a+b):
        arr[i] = 1

    for i in range(a+b, len(arr)):
        arr[i] = 2

nums = [1,0,2,1,0]
better(nums)
print(nums)

# T.C - O(N)
# S.C - O(1)
```
