Brute

```python
def my_brute(arr):

    n = len(arr)
    for i in range(n):
        cnt = 0
        for j in range(n):
            if arr[j] == arr[i]:
                cnt+=1
        
        if cnt > (n // 2):
            return arr[i]

    return -1


arr = [2, 2, 1, 1, 1, 2, 2]
my_brute(arr)

# T.C - O(N^2)
# S.C - O(1)
```
