Brute

```python
def my_brute(arr, k):
    n = len(arr)
    for i in range(n):
        for j in range(i+1, n):
            if arr[i] + arr[j] == k:
                return True
    return -1


arr = [2,6,5,8,11]
k = 14
print(my_brute(arr, k))

# T.C - O(N^2)
# S.C - O(1)
```

Better

```python
def two_sum_exists(arr, target):
    mp = {}
    for i, num in enumerate(arr):
        complement = target - num
        if complement in mp:
            return "YES"
        
        mp[num] = i

    return "NO"


arr = [2,6,5,8,11]
k = 14

print(two_sum_exists(arr, k))
        
```

```python
def two_sum(arr, target):
    mp = {}
    for i, num in enumerate(arr):
        complement = target - num
        if complement in mp:
            return "YES"
        
        mp[num] = i

    return "NO"

arr = [2,6,5,8,11]
k = 14

print(two_sum(arr, k))
```
