Brute force

```python
def brute(arr):
    n = len(arr)
    if n == 0 or n==1:
        print("Array has only one elemet or no elements")
        return
    arr.sort()

    return arr[-2]

arr  = [1,2,3,4,5]
brute(arr)

# T.C - O(N log N)
# S.c - O(1)
```

Optimal 

```python
def my_brute(arr):
    largest = float('-inf')
    second_largest = float('-inf')

    n = len(arr)

    for i in range(n):
        if arr[i] > largest:
            second_largest = largest
            largest = arr[i]

        elif arr[i] > second_largest and arr[i] != largest:
            second_largest = arr[i]

    return second_largest


arr = [1,2,3,4,5]
my_brute(arr)

# T.C - O(N)
# S.C - O(1)
```
