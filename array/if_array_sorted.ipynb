brute

```python
def brute(arr):
    n = len(arr)
    for i in range(n):
        for j in range(i+1, n):
            if arr[j] < arr[i]:
                return False
            
    return True


arr = [1,2,3]
brute(arr)

# T.C - O(N^2)
# S.C - O(1)
```

Optimal

```python
def my_brute(arr):
    n = len(arr)
    cnt = 0
    for i in range(0, n-1):
        if arr[i] > arr[i+1]:
            cnt +=1

    if cnt >=1:
        return False
    else:
        return True

arr = [1,2,4]
my_brute(arr)
```

```python
def optimal(arr):
    n = len(arr)

    for i in range(1, n):
        if arr[i] < arr[i-1]:
            return False
    return True

arr = [1,2,4]
optimal(arr)

# T.C - O(N)
# S.C - O(1)
```
