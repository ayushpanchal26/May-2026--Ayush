```python
def brute(arr):
    cnt = 0 
    n = len(arr)

    for i in range(n):
        num = arr[i]
        count = 0 

        for j in range(n):
            if arr[j] == num:
                count +=1
        
        if count == 1:
            return num 
        
    return -1 

arr = [1,1,2,2,3,4,4,5,5]
brute(arr)
```
