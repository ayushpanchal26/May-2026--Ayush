Brute

```python
def my_brute(arr, k):
    maxi = 0 
    n = len(arr)
    for i in range(0, n-1):
        for j in range(i, n-1):
            currentsum = 0 
            for l in range(i, j+1):
                currentsum += arr[i]
            
            if currentsum == k:
                maxi = max(maxi, j - i+1)
            
    return maxi

nums = [10, 5, 2, 7, 1, 9]
my_brute(nums, 15)

        
```
