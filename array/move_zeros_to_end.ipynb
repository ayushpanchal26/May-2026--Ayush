```python
def brute(arr):
    n = len(arr)
    temp = [0] * n
    index = 0 

    for num in arr:
        if num != 0:
            temp[index] = num
            index+=1

    for i in range(n):
        arr[i] = temp[i]

    return arr

arr = [1,0,2,0,3,0]
brute(arr)

# T.C - O(N)
# S.C - O(N)
```

```python
def optimal(arr):
    j = -1

    for i in range(len(arr)):
        if arr[i] == 0:
            j = i 
            break

    if j == -1:
        return
    
    for i in range(j + 1, len(arr)):
        if arr[i] !=0:
            arr[i], arr[j] = arr[j], arr[i]
            j +=1


nums = [0, 1, 0, 3, 12]
optimal(nums)
print(" ".join(map(str, nums)))

# T.C - O(N)
# S.C - O(1)
```
