Brute

```python
def my_brute(arr):
    n = len(arr)
    for i in range(n):
        num = arr[i]
        cnt = 0
        for j in range(n):
            if arr[j] == num:
                cnt+=1
        if cnt == 1:
            return num
     
    return -1

arr = [4,1,2,1,2]
my_brute(arr)
```

```python
def brute(arr):
    n = len(arr)
    for i in range(n):
        num = arr[i]
        cnt = 0 
        for j in range(n):
            if arr[j] == num:
                cnt+=1
        if cnt == 1:
            return num
        
    return -1

arr = [1,2,3,4,2,1,2,3]
brute(arr)

# T.C - O(N*N)
# S.C - O(1)
```

Better

```python
def better(arr):
    n = len(arr)
    maxi = max(arr)
    hash_arr = [0] * (maxi + 1)

    for num in arr:
        hash_arr[num] +=1
    for num in arr:
        if hash_arr[num] == 1:
            return num
    return -1 

arr = [4,1,2,1,2]
better(arr)

# T.C - O(N) + O(N) + O(N)
# S.C - O(maxElement+1)
```

otpimal

```python
def optimal(arr):
    xorr = 0 
    
    for num in arr:
        xorr ^= num
    return xorr

arr = [4,1,2,1,2]
optimal(arr)
```
