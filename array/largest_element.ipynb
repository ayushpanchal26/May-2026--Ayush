Brute force methods

```python
def largest1(arr):
    arr.sort()

    return arr[-1]


arr = [1,2,3,4,5]
largest1(arr)

# T.C - O(N log N)
# S.C - O(1)
```

Optimal

```python
def largest(arr):
    largest = arr[0]
    n = len(arr)
    for i in range(n):
        if arr[i] > largest:
            largest = arr[i]

    return largest


arr = [1,2,3,4,5]
largest(arr)
```
