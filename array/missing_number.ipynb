```python
def my_brute(arr):
    n = len(arr)
    temp = arr
    missing_num = 0

    for i in range(1, n):
        if i+1 not in temp:
            missing_num = i+1

    return missing_num

arr = [1,2,4,6]
print(my_brute(arr))

    
```

```python
def brute(arr):
    n = len(arr) + 1

    for i in range(1, n+1):
        found = False
        for j in range(n - 1):
            if arr[j] == i:
                found = True
                break 

        if not found:
            return i 
    return -1


arr = [1,2,4,6]
print(brute(arr))
```

```python
def using_hash(arr):
    n = len(arr)
    hash = [0] * (n+1)

    for i in range(n-1):
        hash[arr[i]] += 1
    
    for i in range(1, n+1):
        if hash[i] == 0:
            return i 
        return -1
    
```
