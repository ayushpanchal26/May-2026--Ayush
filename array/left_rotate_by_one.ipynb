Brute

```python
def mybrute(arr):
    arr2 = []
    n = len(arr)

    for i in range(1, n):
        arr2.append(arr[i])

    arr2.append(arr[0])

    return arr2

arr = [1,2,3,4,5]
mybrute(arr)
```

```python
def brute(arr):
    n = len(arr)
    arr2 = [0] * n
    
    for i in range(1, n):
        arr2[i - 1] = arr[i]
    arr2[n-1] = arr[0]

    return arr2

arr = [1,2,3,4,5]
brute(arr)

# T.C - O(N)
# S.C - O(N)
```

Optimal

```python
def optimal(arr):
    temp = arr[0]

    for i in range(1, len(arr)):
        arr[i-1] = arr[i]
    arr[-1] = temp

    return arr

arr = [1,2,3,4,5]
optimal(arr)

# T.C - O(N)
# S.C - O(1)
```
