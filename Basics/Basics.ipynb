```python
# counts digits in a number

def countdigit(n):
    count = 0
    while n>0:
        count +=1 
        n = n//10
    return count

if __name__ == "__main__":
    N = 12345
    print(f"Number: {N}")
    digit = countdigit(N)
    print(f"Number of Digits: {digit}")


# Time Complexity = O(log10N + 1)
# Space Complexity = O(1)
```

```python
# reverse a number 

def reverse_number(n):
    rev = 0 
    while n>0:
        digit = n%10
        rev = rev*10 + digit
        n = n// 10
    return rev

if __name__ == "__main__":
    n = 123
    print(f"Number before reverse: {n}")
    rev = reverse_number(n)
    print(f"Number after reverse {rev}")

# TC = O(log10n)
# SC = O(1)
```

```python
# palindrome number

def palindrome(n):
    rev = 0
    dup = n
    
    while n>0:
        digit = n%10
        rev = rev*10 + digit
        n = n//10

    if rev == dup:
        print("Palindrome")
    else:
        print("Not Palindrome")
    
if __name__ == "__main__":
    num = 121
    palindrome(num)
```

```python
# GCD - Brute force 

def gcd_brute(n1, n2):
    gcd = 1

    for i in range(1, min(n1, n2)):
        if n1%i ==0 and n2%i == 0:
            gcd = i

    return gcd

if __name__ == "__main__":
    n1, n2 = 9, 12
    gcd = gcd_brute(n1, n2)
    print(f"GCD of {n1} and {n2} = {gcd}")


# T.C = O(min(N1, N2))
# S.C = O(1)
```

```python
# GCD - little better

def gcd_little_better(n1, n2):

    for i in range(min(n1, n2), 0, -1):
        if n1%i ==0 and n2%i ==0:
            return i
        
    return 1
        

if __name__ == "__main__":
    n1, n2 = 9, 12
    print(gcd_little_better(n1, n2))

# T.C = O(min(N1, N2))
# S.C = O(1)
        
```

```python
# GCD - Optimal (Using Euclidean ALgorithm)

def gcd_optimal(n1, n2):

    while n1>0 and n2>0:
        if n1>n2:
            n1 = n1%n2
        else:
            n2 = n2%n1
    
    if n1 == 0:
        return n2
    
    return n1

if __name__ == "__main__":
    n1, n2 = 9, 12
    print(gcd_optimal(n1, n2))


# T.C = O(min(N1, N2))
# S.C = O(1)
```

```python
# Armstrong 

def armstrong(num):
    k = len(str(num))
    sum = 0
    n = num
    while n>0:
        ld = n%10
        sum = sum+ld **k
        n = n//10

    return sum == num


if __name__ == "__main__":
    n = 153
    print(armstrong(n))

# T.C = O(log10N + 1)
# S.C = O(1)
```

```python
# Print all Divisors of a given number

# Brute Force
def all_divisor(n):
    list1 = []
    
    for i in range(1, n+1):
        if n%i == 0:
            list1.append(i)

    return list1


if __name__ == "__main__":
    n = 36
    print(all_divisor(n))

# T.C = O(N)
# S.C = O(N)
```

```python
# Better approach

import math


def all_div_better(n):
    list1 = []

    for i in range(1, int(math.isqrt(n))+1):

        if n%i ==0:
            list1.append(i)

            if i!= n//i:
                list1.append(n//i)

    return list1

if __name__ == "__main__":
    n = 36
    store = all_div_better(n)
    store.sort()
    print(store)
```

```python
# Prime Number 
# Brute force

def prime(n):
    count = 0

    for i in range(1, n+1):
        if n%i == 0:
            count+=1
        

    if count>2:
        print("NOT PRIME")
    else: 
        print("PRIME")

    return 1 


if __name__ == "__main__":

    n = 8
    print(prime(n))
```
