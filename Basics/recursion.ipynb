```python
# Print Name using Recusrion N times. 

def rec_name(i, n):
    if i>n:
        return
    print("Object")
    rec_name(i+1, n)

rec_name(1,8)

# T.C - O(N)
# S.C - O(N)
```

```python
# Print 1 to N using Recursion

def one_to_n(i, n):
    if i>n:
        return
    print(i)
    one_to_n(i+1, n)

one_to_n(1, 10)
```

```python
def n_to_one(i, n):
    if n<i:
        return
    print(n)
    n_to_one(i, n-1)

n_to_one(1, 10)
```

```python
def sum_of_first_n_no(n, sum):
    if n < 1:
        print(sum)
        return 
    
    sum_of_first_n_no(n-1, sum+n)

sum_of_first_n_no(3, 0)
```

```python
def factorial(n, fact):
    if n<=1:
        print(fact)
        return
    factorial(n-1, fact*n)

factorial(5, 1)
```

```python
# REVERSE AN ARRAY
def print_array(arr):
    print("REVERSER ARRAY!!")

    for i in arr:
        print(i, end=' ')

def reverse_array(arr):
    p1 = 0
    p2 = len(arr) - 1
    while p1<p2:
        arr[p1], arr[p2] = arr[p2], arr[p1]
        p1 +=1
        p2 -=1

    print_array(arr)

arr = [5,4,3,2,1]
reverse_array(arr)
```

```python
def print_array(arr):
    print("REVERSER ARRAY!!")

    for i in arr:
        print(i, end=' ')

def recursion_reverse_array(start, arr, end):
    if start>end:
        return
    arr[start], arr[end] = arr[end], arr[start]
    recursion_reverse_array(start+1, arr, end-1)

arr = [5,4,3,2,1]
n = len(arr)
recursion_reverse_array(0, arr, n-1)
print_array(arr)
```

```python
def is_palindrome(i, string):
    lenght = len(string)
    if i>=len(string)/2:
        return True
    if string[i] != string[lenght-i-1]:
        return False
    
    return is_palindrome(i+1, string)


string1 = "MADAM"
is_palindrome(0, string1)

# T.C - O(N)
# S.C - O(1)
```

```python
def fibonaci(n):
    a = 0
    b = 1
    if n == 1:
        print(a)
    elif n == 2:
        print(a, b)

    else:
        print(a, b, end= " ")
        for i in range(n-2):
            c = a+b
            a = b
            b = c
            print(b, end=" ")

n = 4
fibonaci(n)
```

```python
def fibo_recusrion(n):
    if n == 0:
        return 0
    
    if n == 1:
        return 1
    
    return fibo_recusrion(n-1) + fibo_recusrion(n-2)


for i in range(10):
    print(fibo_recusrion(i), end=' ')
```

```python
def ffb(n):
    if n == 0:
        return 0 
    
    if n == 1:
        return 1
    
    return ffb(n-1) + ffb(n-2)

for i in range(4):
    print(ffb(i), end=' ')
```
