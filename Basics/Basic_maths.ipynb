Count All Digit

```python
# 1. Count all digits of a number - brute

def count_digit(n):
    n_range = str(n)
    count = 0
    for i in range(len(n_range)):
        count+=1

    return count

n = 124
print(count_digit(n))
```

```python
# Brute 2

def count_digit_(n):
    cnt = 0 

    while n>0:
        cnt +=1

        n = n//10

    return cnt

n = 1234
print(count_digit_(n))

# T.C - O(log10N + 1),
# S.C - O(1)
```

Reverse a Number

```python
def rev_(n):

    rev = 0
    while n>0:
        digit = n%10
        rev = rev*10 + digit
        n = n//10

    return rev

n = 123456789
print(rev_(n))

# T.C = O(Log10N)
# S.C = O(1)
```

Palindrome

```python
def palindrome(n):

    rev = 0
    same = n

    while n > 0:

        digit = n%10
        rev = rev*10 + digit
        n = n//10

    if rev == same:
        return True
    return False

n = 121
print(palindrome(n))
```

GCD of Two Numbers

```python
def gcd_brute(n1, n2):
    gcd = 0 

    for i in range(1, min(n1, n2)):
        if n1 % i == 0 and n2 % i == 0:
            gcd = i

    return gcd


n1 = 9
n2 = 12

print(gcd_brute(n1, n2))

# T.C - O(min(n1, n2))
# S.C - O(1)
```

```python
def gcd_better(n1, n2):
    
    for i in range(min(n1, n2), 1, -1):
        if n1%i ==0 and n2%i ==0:
            return i
    return 1

n1 = 9
n2 = 12

print(gcd_better(n1, n2))

# T.C - O(min(n1, n2))
# S.C - O(1)
```

```python
def gcd_optimal(n1, n2):

    while n1> 0 and n2>0:
        if n1>n2:
            n1 = n1%n2
        else:
            n2 = n2%n1

    if n1 == 0:
        return n2
    return n1

n1 = 20
n2 = 15

print(gcd_optimal(n1, n2))
# T.C = O(min(N1, N2))
# S.C = O(1)
```

Armstrong Number

```python
def armstrong(n):
    k = len(str(n))
    copy = n
    sum = 0

    while copy>0:
        ld = copy %10
        sum += ld ** k
        copy = copy // 10

    return sum == n

n = 153
print(armstrong(n))
```

All divisors

```python
def all_divisor_brute(n):
    list1 = []
    for i in range(1, n+1):
        if n%i == 0:
            list1.append(i)

    return list1

n = 6
print(all_divisor_brute(n))

# T.C - O(N)
# S.C - O(N)
```

```python
import math
def all_divisor_optimal(n):
    list1 = []
    for i in range(1, int(math.isqrt(n)) +1 ):
        if n%i == 0:
            list1.append(i)

            if i!=n//i:
                list1.append(n//i)

    return list1

n = 6
print(all_divisor_optimal(n))

# T.C - O(sqrt(N))
# S.C - O(2*sqrt(N))
```

Prime Number

```python
def prime(n):
    cnt = 0
    for i in range(1, n+1):
        if n%i == 0:
            cnt+=1

    if cnt <= 2:
        return True
    else:
        return False

n = 1
print(prime(n))

# T.C - O(N)
# S.C - O(1)
```
