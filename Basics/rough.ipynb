```python
def countss(n):
    n_range = str(n)
    counts = 0 
    for i in range(len(n_range)):
        counts+=1

    return counts

n = 12344
print(countss(n))
```

```python
def count2(n):

    cnt = 0 
    while n>0:
        cnt +=1

        n = n//10

    return cnt

n = 12345
print(count2(n))
```

```python
import math

def count3(n):
    cnt = int(math.log10(n) + 1)
    return cnt

n = 12345
print(count3(n))
```

```python
def reverse(n):
    rev = 0

    while n>0:
        ld = n%10
        rev = rev*10 + ld
        n = n//10

    return rev

n = 12345
print(reverse(n))

# T.C -  O(log10N)
# S.C - O(1)
```

```python
def palindrome(n):
    rev = 0
    org = n 

    while n>0:
        ld = n %10
        rev = rev*10 + ld
        n = n//10

    if rev == org:
        return True
    return False

n = 1221
print(palindrome(n))
```

```python
def gcd1(n1, n2):
    gcd = 0
    for i in range(1, min(n1, n2)):
        if n1%i == 0 and n2%i == 0:
            gcd = i

    return gcd

n1 = 9
n2 = 12
print(gcd1(n1, n2))

```

```python
def gcd2(n1, n2):
    gcd = 0
    for i in range(min(n1, n2), 1, -1):
        if n1%i == 0 and n2%i == 0:
            gcd = i

    return gcd

n1 = 9
n2 = 12
print(gcd2(n1, n2))
```

```python
def gcd3(n1, n2):
    while n1 > 0 and n2> 0:
        if n1%n2: 
            n1 = n1%n2
        else:
            n2 = n2 %n1 
    if n1 == 0:
        return n2
    return n1
```

```python
def one_to_n(i, n):
    if i>n:
        return
    
    print(i)
    one_to_n(i+1, n)

one_to_n(1, 10)
```

```python
def n_to_one(i, n):
    if n<i:
        return
    print(n)
    n_to_one(i, n-1)

n_to_one(1, 10)
```

```python
def sum_of_first(n, sum):
    if n<1:
        print(sum)
        return 
    
    sum_of_first(n-1, sum+n)

sum_of_first(3, 0)
```

```python
def factorial(n, fact):
    if n<=1:
        print(fact)
        return
    
    factorial(n-1, fact*n)
```

```python
def print_array(arr):
    for i in arr:
        print(i, end=' ')

def reverse_array(start, arr, end):
    if start>end:
        return
    
    arr[start], arr[end] = arr[end], arr[start]
    reverse_array(start+1, arr, end-1)


arr = [5,4,3,2,1]
n = len(arr)

reverse_array(0, arr, n-1)
print_array(arr)
```

```python
def is_palindrome(i, string):
    length = len(string)

    if i > length/2:
        return True
    
    if string[i] != string[length-i-1]:
        return False
    
    return is_palindrome(i+1, string)

is_palindrome(0, 'MADAM')
```

```python
def fibonacci(n):
    if n == 0 :
        return 0 
    
    if n == 1:
        return 1
    
    return fibonacci(n-1) + fibonacci(n-2)


for i in range(5):
    print(fibonacci(i), end=' ')
```
