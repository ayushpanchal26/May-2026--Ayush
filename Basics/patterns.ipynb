```python
n = 5
spaces = 2*n-1

for i in range(1, n+1):

    for j in range(1, i+1):
        print(j, end="")

    for j in range(1, spaces+1):
        print(" ", end="")

    for j in range(i, 0, -1):
        print(j, end="")

    print()
    spaces -=2

```

```python
n = 5

for i in range(n):
    print(" " * (n-i-1), end="")

    ch = ord('A')
    breakpoint = (2*i+1) // 2

    for j in range(1, 2*i+2):
        print(chr(ch), end="")

        if j<=breakpoint:
            ch+=1
        else:
            ch-=1

    print()
```
